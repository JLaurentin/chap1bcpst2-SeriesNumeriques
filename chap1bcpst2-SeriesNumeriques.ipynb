```python
import numpy as np
from math import factorial
import matplotlib.pyplot as plt
```

## Représentation d'une suite numérique (u_n)  :

```python
print('Entrer le terme général de la suite : ')
uk = input('u_k vaut : ')
k0 = int(input('le premier indice k0 vaut : '))
```

```python
def calculTermesU(uk, k0, n):
    return [eval(uk) for k in range(k0, n+1)]
```

```python
n = 10
U = calculTermesU(uk, k0, n)
print(U)
```

```python
xn = range(k0,n+1)
plt.plot(xn, U, 'o-.')
plt.xlim(left=k0-0.2,right=n+0.2)
plt.title('Représentation des termes de la suite u')
plt.show()
```

## Représentation des termes de la suite $(S_n)$ où $S_n = \displaystyle\sum_{k=k_0}^n u_k$

### 1. *Si la liste U a déjà été construite :*

```python
def sommePartielle1(U, k0 ,n):
    """Calcul de (S_{k0},S_{k0+1},...,S_n) connaissant U = (u_{k0},u_{k0+1},...,u_n)"""
    Sp = [U[0]] # unitialisation avec Sp = [u_{k0}]
    for k in range(1, n-k0+1): # il y a (n-k0) termes à calculer
        Sp.append(Sp[-1]+U[k])
    return Sp
```

```python
def sommePartielle2(U,k0,n):
    """ pour une forme plus condensée de la fonction précédente..."""
    return [sum(U[:k+1]) for k in range(n-k0+1)]
```

```python
n = 10
Sp1 = sommePartielle1(U,k0,n)
plt.plot(xn, Sp1, 'ro-', label='Somme partielle')
plt.vlines(xn, 0, U, 'b', label='suite $(u_n)$')
plt.legend(loc='lower right')
plt.show()
```

### 2. *Si on veut déterminer en même temps U et Sp connaissant l'expression de uk :*

#### a) Exemple de la série géométrique de terme général uk = q**k pour tout k>=k0

```python
def sommePartielleGeom(q, k0, n):
    """
    renvoie la liste U des termes de la suite (un)=(q**n)
    et la liste Sp des termes de la somme partielle.
    """ 
    U, Sp = [], []
    S = 0
    for k in range(k0, n+1):
        u = k*q**k
        S += u
        U.append(u)
        Sp.append(S)
    return U, Sp
```

```python
q = 1/2
k0 = 1
n = 10
U, Sp=sommePartielleGeom(q, k0, n)
```

```python
def traceSommePartielle(U, Sp, k0, n):
    xn = range(k0, n+1)
    plt.plot(xn, Sp, 'ro', markersize=5, label='Somme partielle')
    plt.vlines(xn, 0, U, 'b', lw=3, alpha=0.5, label='suite (u_n)')
    plt.legend(loc='lower right')
    plt.xlim(left=k0-0.2,right=n+0.2)
    plt.ylim(bottom=0,top=Sp[-1]+0.1)
    plt.show()
    
traceSommePartielle(U,Sp,k0,n) 
```

#### b) Exemple de la série exponentielle de terme général $u_k = x^k/k!$ pour tout $k\geq 0$

```python
def sommePartielleExponentielle(x, n):
    """ renvoie la liste des termes de la suite (un)=(x**n/factorial(n)) et la liste Sp des termes de la somme partielle."""
    U, Sp = [], []
    S = 0
    for k in range(n+1):
        u = x**k/factorial(k)
        S += u
        U.append(u)
        Sp.append(S)
    return U, Sp
```

```python
x = 1
k0 = 0
n = 10
U,Sp=sommePartielleExponentielle(x, n)
traceSommePartielle(U, Sp, k0, n) 
```

#### c) Exemple de la série harmonique de terme général $u_k = 1/k$ pour tout $k\geq 1$

```python
def sommePartielleHarmonique(n):
    """
    on supposera la suite définie à partir de n = 1
    renvoie la liste U des termes de la suite (un)=(1/n)
    et la liste Sp des termes de la somme partielle.    
    """
    U, Sp = [], []
    S = 0
    for k in range(1, n+1):
        u = 1/k
        S += u
        U.append(u)
        Sp.append(S)
    return U, Sp
```

```python
n = 20
U,Sp=sommePartielleHarmonique(n)
k0 = 1
traceSommePartielle(U, Sp, k0, n) 
```
